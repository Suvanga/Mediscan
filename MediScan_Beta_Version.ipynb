```python
import torch
from torch.utils.data import Dataset
from PIL import Image
import os

#writing a custom dataset class
class ChestXrayDataset(Dataset):
    def __init__(self, dataframe, root_dir, transform=None):
      #Initialize our custom data set
        self.labels_frame = dataframe
        self.root_dir = root_dir
        self.transform = transform

       #classification is going to be done with these labels
        self.all_labels = [
            'Atelectasis', 'Cardiomegaly', 'Effusion', 'Infiltration', 'Mass', 'Nodule',
            'Pneumonia', 'Pneumothorax', 'Consolidation', 'Edema', 'Emphysema',
            'Fibrosis', 'Pleural_Thickening', 'Hernia', 'No Finding'
        ]

    def __len__(self):
        return len(self.labels_frame)

    def __getitem__(self, idx):
        # 1. Get Image Path
        img_name = os.path.join(self.root_dir, self.labels_frame.iloc[idx, 0])
        image = Image.open(img_name).convert('RGB')

        # 2. Get Labels (Multi-Hot Encoding)
        label_str = self.labels_frame.iloc[idx, 1]

        # Check against ALL 15 labels
        label_vec = torch.tensor(
            [1.0 if disease in label_str else 0.0 for disease in self.all_labels],
            dtype=torch.float32
        )

        # 3. Transform
        if self.transform:
            image = self.transform(image)

        return image, label_vec
```

```python
import torch
import torch.nn as nn
from torchvision import models

#Device agnostic code
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
class ModelV1(nn.Module):
    """
    ModelV1: A wrapper for DenseNet121 using Transfer Learning.
    Args:
        output_shape (int): The number of diseases to predict.
    """
    def __init__(self, output_shape: int):
        super().__init__()

        # 1. Load the Pretrained Model (The "Brain")
        self.densenet = models.densenet121(weights='DEFAULT')

        # 2. Freeze the "Body" (Feature Extractor)
        # We don't want to destroy the patterns it learned from ImageNet.
        for param in self.densenet.features.parameters():
            param.requires_grad = False

        # 3. Replace the "Head" (Classifier)
        # DenseNet121's feature extractor outputs 1024 features.

        self.densenet.classifier = nn.Linear(in_features=1024, out_features=output_shape)

    def forward(self, x):
        return self.densenet(x)

```

```python
import torch.nn.functional as F
def train_step(model: torch.nn.Module,
               dataloader : torch.utils.data.DataLoader,
               loss_fn: torch.nn.Module,
               optimizer: torch.optim.Optimizer,
               device = device):

  model.train()

  train_loss, train_acc = 0, 0

  for batch, (X,y) in enumerate(dataloader):
    X,y = X.to(device), y.to(device)


    if y.ndim == 1:
            y = F.one_hot(y.long(), num_classes=15).float()
    else:
            y = y.float()

    #Do the forward pass
    y_pred = model(X) #This outputs model logits

    #Calculate the loss
    loss = loss_fn(y_pred, y)
    train_loss += loss.item()

    #Optimizer zero grad
    optimizer.zero_grad()

    #loss backward
    loss.backward()

    #Optimizer step
    optimizer.step()

    # Calculate multi-label accuracy
    y_prob = torch.sigmoid(y_pred) # Convert logits to probabilities
    y_pred_binary = (y_prob > 0.5).float() # Threshold probabilities to get binary predictions
    # Calculate element-wise accuracy over all labels and samples in the batch
    train_acc += (y_pred_binary == y).float().mean().item()


  train_loss = train_loss / len(dataloader)
  train_acc = train_acc/len(dataloader)
  return train_loss, train_acc

```

```python
import torch.nn.functional as F
def test_step(model: torch.nn.Module,
              dataloader : torch.utils.data.DataLoader,
              loss_fn: torch.nn.Module,
              device = device):
    with torch.inference_mode():
        model.eval()
        test_loss, test_acc, test_recall = 0, 0, 0

        for batch, (X,y) in enumerate(dataloader):
            X,y = X.to(device), y.to(device)

            if y.ndim == 1:
                y = F.one_hot(y.long(), num_classes=15).float()
            else:
                y = y.float()

            # Do the forward pass
            test_preds_logits = model(X)
            loss = loss_fn(test_preds_logits, y)
            test_loss += loss.item()

            # 3. Predictions (Sigmoid + Threshold)
            test_prob = torch.sigmoid(test_preds_logits)
            test_pred_binary = (test_prob > 0.5).float()

            # 4. Accuracy
            test_acc += (test_pred_binary == y).float().mean().item()

            # 5. RECALL FORMULA
            # Logic: (True Positives) / (All Actual Positives)
            # We add 1e-8 to avoid dividing by zero if there are no diseases in the batch
            true_positives = ((test_pred_binary == 1) & (y == 1)).float().sum()
            actual_positives = (y == 1).float().sum()
            test_recall += (true_positives / (actual_positives + 1e-8)).item()

        # Average over batches
        test_loss = test_loss / len(dataloader)
        test_acc = test_acc / len(dataloader)
        test_recall = test_recall / len(dataloader)

        return test_loss, test_acc, test_recall
```

```python
def train_test_function(model: torch.nn.Module,
                        train_dataloader: torch.utils.data.DataLoader,
                        test_dataloader: torch.utils.data.DataLoader,
                        optimizer: torch.optim.Optimizer,
                        loss_fn,
                        Epochs: int = 5,
                        device=device):

    # Initialize results dictionary
    results = {
        "train_loss": [],
        "train_acc": [],
        "test_loss": [],
        "test_acc": [],
        "test_recall": []
    }

    from timeit import default_timer as timer
    from tqdm.auto import tqdm

    for epoch in tqdm(range(Epochs)):
        train_time_start = timer()


        train_loss, train_acc = train_step(model=model,
                                           dataloader=train_dataloader,
                                           loss_fn=loss_fn,
                                           optimizer=optimizer,
                                           device=device)

        train_time_end_ep1 = timer()
        print(f"Epoch {epoch} for training is finished at {train_time_end_ep1-train_time_start}")
        test_loss, test_acc, test_recall = test_step(model=model,
                                                     dataloader=test_dataloader,
                                                     loss_fn=loss_fn,
                                                     device=device)



        print(f"\nEpoch: {epoch+1} | "
              f"train_loss: {train_loss:.4f} | "
              f"train_acc: {train_acc:.4f} | "
              f"test_loss: {test_loss:.4f} | "
              f"test_acc: {test_acc:.4f} | "
              f"Recall: {test_recall:.4f}")

        results["train_loss"].append(train_loss)
        results["train_acc"].append(train_acc)
        results["test_loss"].append(test_loss)
        results["test_acc"].append(test_acc)
        results["test_recall"].append(test_recall)

    return results
```

```python
# 2. Define Model V2 (The Specialist)
class ModelV2(nn.Module):
    """
    ModelV2: An upgraded DenseNet121 with Fine-Tuning and Dropout.
    Args:
        output_shape (int): The number of diseases to predict.
    """
    def __init__(self, output_shape: int):
        super().__init__()

        # Load the Pretrained Brain
        self.densenet = models.densenet121(weights='DEFAULT')

        # UPGRADE 1: Fine-Tuning
        # We freeze the early layers (simple edges/shapes)
        for param in self.densenet.features.parameters():
            param.requires_grad = False

        # ... BUT we UNFREEZE the last "Dense Block".
        # This allows the model to learn high-level X-ray concepts (haze, consolidation).
        for param in self.densenet.features.denseblock4.parameters():
            param.requires_grad = True
        for param in self.densenet.features.norm5.parameters():
            param.requires_grad = True

        # UPGRADE 2: A Deeper, Safer Head
        # Old Head: 1024 -> 15
        # New Head: 1024 -> 512 -> ReLU -> Dropout -> 15
        self.densenet.classifier = nn.Sequential(
            nn.Linear(1024, 512),            # Squeeze features down
            nn.ReLU(),                       # Activation
            nn.Dropout(p=0.2),               # Randomly drop 20% to prevent memorization
            nn.Linear(512, output_shape)     # Final prediction
        )

    def forward(self, x):
        return self.densenet(x)

# 3. Instantiate Model B
modelB = ModelV2(output_shape=15).to(device)

print(f"✅ ModelV2 (modelB) created on {device}!")
print("Last block is unfrozen for fine-tuning.")
```

```python
import torch
from torchvision import transforms
from PIL import Image
import matplotlib.pyplot as plt

# 1. SETUP: Define the 15 Classes (Must match your training order exactly)

class_names = [
    'Atelectasis',
    'Cardiomegaly',
    'Consolidation',
    'Edema',
    'Effusion',
    'Emphysema',
    'Fibrosis',
    'Hernia',
    'Infiltration',
    'Mass',
    'No Finding',
    'Nodule',
    'Pleural_Thickening',
    'Pneumonia',
    'Pneumothorax'
]

# 2. LOAD MODEL: Initialize and load weights
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = ModelV2(output_shape=15) # Re-create the structure
model.load_state_dict(torch.load("MediScan_V2_Balanced.pth", map_location=device)) # Load the brain
model.to(device)
model.eval() # Set to evaluation mode (turns off training layers)

# 3. PREPARE IMAGE: Define the same transforms as training (Resize + Normalize)
predict_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

def predict_step(model, image_tensor):
    """The 'Inference' version of your test_step."""
    model.eval()
    with torch.inference_mode():
        # Do the forward pass
        logits = model(image_tensor)

        # Apply Sigmoid to get independent probabilities (0.0 to 1.0)
        probs = torch.sigmoid(logits)

    return probs

def predict_single_image(image_path):
    try:
        image = Image.open(image_path).convert('RGB')

        # Show the image
        plt.imshow(image)
        plt.axis('off')
        plt.show()

        # Transform image to tensor
        image_tensor = predict_transform(image).unsqueeze(0).to(device)

        # USE THE STEP FUNCTION HERE
        probs = predict_step(model, image_tensor)[0] # [0] because it's a batch of 1

        print("\n--- 🔍 MediScan Analysis ---")
        results = []
        for i, prob in enumerate(probs):
            results.append((class_names[i], prob.item() * 100))

        # Sort by highest confidence
        results.sort(key=lambda x: x[1], reverse=True)

        for name, score in results:
            print(f"{name}: {score:.2f}%")

    except Exception as e:
        print(f"Error: {e}")
```

```python
import torch
import numpy as np

# 1. Load the "Paranoid" Model (ModelV2)
# Ensure the filename matches EXACTLY what you have on disk
model_path = "MediScan_V2_Balanced.pth"  # Check casing!

try:
    modelB = ModelV2(output_shape=15) # Re-initialize structure
    modelB.load_state_dict(torch.load(model_path, map_location=device))
    modelB.to(device)
    modelB.eval() # Important: Set to evaluation mode
    print(f"✅ Loaded {model_path} successfully.")
except FileNotFoundError:
    print(f"❌ ERROR: Could not find '{model_path}'. Please verify the filename.")

# 2. THE INSTANT FIX: Manually adjust the 'Bias'
# This tells the model: "Start with a negative bias against rare diseases."
# It forces the model to need STRONG evidence before predicting Hernia.

# A. Calculate the 'Prior Probability' (How rare is each disease?)
positive_counts = df.iloc[:, 1].str.get_dummies(sep='|').sum()
total_samples = len(df)
probs = positive_counts / total_samples

# B. Calculate the Bias Correction
# Math: log(P / (1 - P)) is the "natural" bias for a dataset
# We act aggressively on rare classes (Low P -> Very Negative Bias)
bias_correction = np.log(probs / (1 - probs))
bias_correction = torch.tensor(bias_correction.values).float().to(device)

# C. Apply it to the Final Layer
# We look for the last classifier layer. In DenseNet121, it's usually:
# model.densenet.classifier (which is a Sequential block in your V2)
# The final Linear layer is at index [3] of that sequence.
with torch.no_grad():
    final_layer = modelB.densenet.classifier[3]

    # We mix the old learned bias with our new correction
    # 50% learned + 50% correction usually stabilizes the predictions perfectly
    final_layer.bias.data = final_layer.bias.data * 0.5 + bias_correction * 0.5

print("✅ Instant Calibration Complete!")
print("   - Hernia/Pneumothorax biases have been lowered.")
print("   - Common diseases remain mostly unchanged.")

# 3. Save as V3
torch.save(modelB.state_dict(), "MediScan_V3_Calibrated_Manual.pth")
print("✅ Saved as 'MediScan_V3_Calibrated_Manual.pth'")
```

```python
import torch
import random
import matplotlib.pyplot as plt
import numpy as np

# 1. Force Define Device (Mac M1/M2/M3 specific)
if torch.backends.mps.is_available():
    device = torch.device("mps")
    print("✅ Using Apple MPS (GPU) Acceleration")
elif torch.cuda.is_available():
    device = torch.device("cuda")
    print("✅ Using NVIDIA CUDA (GPU)")
else:
    device = torch.device("cpu")
    print("⚠️ Using CPU (Slow)")

# 2. Re-Initialize Model & Load Weights
try:
    # Initialize the structure
    modelB = ModelV2(output_shape=15)

    # Load the weights (Use map_location to be safe)
    state_dict = torch.load("MediScan_V3_Calibrated_Manual.pth", map_location=device)
    modelB.load_state_dict(state_dict)

    # CRITICAL STEP: Move the model to the device
    modelB.to(device)
    modelB.eval()
    print("✅ Model loaded and moved to device successfully.")

except Exception as e:
    print(f"❌ Error loading model: {e}")

# 3. Define the CORRECT Class Names (Alphabetical)
class_names = [
    'Atelectasis', 'Cardiomegaly', 'Consolidation', 'Edema', 'Effusion',
    'Emphysema', 'Fibrosis', 'Hernia', 'Infiltration', 'Mass',
    'No Finding', 'Nodule', 'Pleural_Thickening', 'Pneumonia', 'Pneumothorax'
]

# 4. Run the Visual Check
# We paste the function here to ensure 'device' is used correctly inside it
def check_random_samples_safe(model, dataset, class_names, num_samples=3):
    fig, axes = plt.subplots(num_samples, 1, figsize=(10, 5 * num_samples))
    if num_samples == 1: axes = [axes]

    for i in range(num_samples):
        idx = random.randint(0, len(dataset) - 1)
        image, label = dataset[idx] # Image is a Tensor

        # Move Image to Device (Fixes the RuntimeError)
        image_tensor = image.unsqueeze(0).to(device)

        with torch.no_grad():
            logits = model(image_tensor)
            probs = torch.sigmoid(logits).squeeze()

        # Get Top 3 Predictions
        top3_prob, top3_idx = torch.topk(probs, 3)

        # Format Truth
        true_diseases = [class_names[j] for j, val in enumerate(label) if val == 1]
        if not true_diseases: true_diseases = ["No Finding"]

        # Format Prediction Text
        pred_text = []
        for prob, idx in zip(top3_prob, top3_idx):
            name = class_names[idx.item()]
            score = prob.item() * 100
            pred_text.append(f"{name}: {score:.1f}%")

        # Display
        img_display = image.permute(1, 2, 0).cpu().numpy()
        # Denormalize
        mean = np.array([0.485, 0.456, 0.406])
        std = np.array([0.229, 0.224, 0.225])
        img_display = std * img_display + mean
        img_display = np.clip(img_display, 0, 1)

        axes[i].imshow(img_display)
        axes[i].axis('off')
        axes[i].set_title(
            f"Truth: {', '.join(true_diseases)}\nPreds: {', '.join(pred_text)}",
            loc='left', fontsize=12, color='blue'
        )

    plt.tight_layout()
    plt.show()

# Run it!
check_random_samples_safe(modelB, test_dataset, class_names, num_samples=3)
```
